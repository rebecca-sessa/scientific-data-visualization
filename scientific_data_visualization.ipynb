# Scientific data visualization with Python

This notebook explores basic scientific data analysis and visualization workflows using pandas and seaborn.

```python
#import
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

## Dataset overview

The dataset contains simplified experimental information for hypothetical compounds, including biological activity, solubility and toxicity indicators.

```python
#dataset
data = {
    "Compound": ["Cmpd_A", "Cmpd_B", "Cmpd_C", "Cmpd_D", "Cmpd_E"],
    "Activity_IC50": [12.5, 45.2, 8.7, 60.1, 15.3],
    "Solubility": [85, 40, 92, 25, 70],
    "Toxicity_Score": [2.1, 5.8, 1.9, 7.2, 3.0]
}

df = pd.DataFrame(data)

df
```

```python
#statistic analysis
df.describe()
```

## Interpretation

Compounds with higher solubility tend to show lower IC50 values in this small dataset, suggesting a potential relationship between solubility and biological activity. Further analysis with larger datasets would be necessary to validate this trend.

```python
#graph 1
sns.scatterplot(
    data=df,
    x="Solubility",
    y="Activity_IC50",
    hue="Compound",
    s=120
)

plt.title("Solubility vs Biological Activity")
plt.show()
```

## Correlation analysis

The heatmap highlights relationships between physicochemical and biological variables. Toxicity appears positively associated with IC50 values, while solubility shows an inverse trend.

```python
corr = df.drop(columns="Compound").corr()

sns.heatmap(
    corr,
    annot=True
)

plt.title("Correlation Matrix")
plt.show()
```
